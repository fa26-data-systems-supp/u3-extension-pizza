### 5. Data quality audit

```python
# checking missing artwork title 
%%sql
SELECT COUNT(*) AS missing_titles
FROM artworks
WHERE title IS NULL OR TRIM(title) = '';
```

Found: 39 artworks have missing titles.
Decision: Leave these records unchanged, but add a warning that some titles are missing.
Why: These artworks may still contain useful information. Replacing missing titles with “Untitled” could introduce inaccurate information, so the original values should remain.

```python
# artworks classification
%%sql
SELECT COUNT(*) AS placeholder_classifications
FROM artworks
WHERE LOWER(TRIM(classification)) = '(not assigned)';
```

Found: 632 artworks have (not assigned) as their classification.
Decision: Clean a working copy by converting this placeholder to SQL NULL.
Why: (not assigned) represents missing information rather than an actual artwork category. Converting it to NULL would make missing classifications easier to identify and prevent this placeholder from being treated as a meaningful category.
