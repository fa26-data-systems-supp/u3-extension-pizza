### 1.Introduction

**Driving question:** How does the Museum of Modern Art’s collection differ from four different departments.
    
For this question, we are trying to explore how the number of artworks, the types of artworks, and the creation dates of artworks differ across departments. This question matters because it can help us understand how artworks are distributed across the museum and which time periods are represented in each department. Thus, our findings can help visitors gain a clearer and better understanding of the collections in each department of the museum.
